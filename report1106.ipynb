## 2. loc vs iloc

1. `loc`으로 인덱스 10~15 행의 `sex`, `age` 열을 선택해 출력하세요.
2. `iloc`으로 같은 열(`sex`는 2번째, `age`는 3번째 위치)을 행 위치 10:15로 선택해 출력하세요.
3. 두 결과의 행 수를 각각 출력하고, 왜 다른지 주석으로 쓰세요.

```python
loc_result = titanic.loc[10:15, ["sex", "age"]]
print(loc_result)

iloc_result = titanic.iloc[10:15, 2:4]
print("\n", iloc_result)

print("\nloc_shape :", loc_result.shape)
print("iloc_shape :", iloc_result.shape)

# 두 결과가 다른 이유는 loc와 iloc가 데이터를 접근하는 기준이 각각 이름과 숫자로 다르기 때문에 
# 같은 10 : 15행을 입력했지만 출력되는 행이 다른 것을 확인할 수 있습니다.
```

## 5. 복합 조건

각각 조건을 만족하는 행 수를 출력하세요.

1. 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
2. 1등석 **또는** 요금 100 이상인 승객 (`|`)
3. 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)

```python
first_mask = (titanic["pclass"] == 3) & (titanic["sex"] == "male") & (titanic["alive"] == "yes")
alive_result = titanic[first_mask]
second_mask = (titanic["pclass"] == 1) | (titanic["fare"] >= 100)
fare_result = titanic[second_mask] 
third_mask = (titanic["alone"] == True)
not_alone_result = titanic[~third_mask]

print("1. ", len(alive_result))
print("2. ", len(fare_result))
print("3. ", len(not_alone_result))
```

## 6. isin · between

1. `isin`으로 Cherbourg 또는 Queenstown에서 탑승한 승객을 골라 `shape`를 출력하세요.
2. `between`으로 나이 20~30세(양 끝 포함) 승객을 골라 `shape`와 나이의 min·max를 출력하세요.
3. 같은 범위를 양 끝 미포함(`inclusive="neither"`)으로 바꿔 `shape`를 출력하세요.


```python
isin_mask = titanic["embark_town"].isin(["Cherbourg", "Queenstown"])
first_result = titanic[isin_mask]
print("first_shape :", first_result.shape)

between_mask = titanic["age"].between(20,30)
second_result = titanic[between_mask]
print("second_shape :", second_result.shape)
print("second_min :", second_result["age"].min())
print("second_max :", second_result["age"].max())

inclusive_between_mask = titanic["age"].between(20,30, inclusive="neither")
third_result = titanic[inclusive_between_mask]
print("third_shape :", third_result.shape)
```


## 7. loc[조건, 열]

12살 이하 **그리고** 3등석 승객의 `age`, `sex`, `survived` 열만 `kids` 변수에 담고, 행 수와 생존률(`survived` 평균, 소수 둘째 자리)을 출력하세요.


```python
mask = (titanic["age"] <= 12) & (titanic["pclass"] == 3)
kids = ["age", "sex", "survived"]
result = titanic.loc[mask, kids]
survive_result = round(result["survived"].mean(), 2)

print("len_result :", len(result))
print("survive_result :", survive_result)
```


## 8. query

1. 변수 `min_fare = 50`을 만들고, `query`로 Cherbourg 탑승 **그리고** 요금이 `min_fare` 이상인 승객의 `shape`를 출력하세요.
2. 5-1번 조건(3등석 **그리고** 남성 **그리고** 생존한 승객)을 `query`로 다시 써서 행 수가 같은지 확인하세요.


```python
min_fare = 50
result = titanic.query("embark_town == 'Cherbourg' and fare >= @min_fare")

result_query = titanic.query("pclass == 3 and sex == 'male' and alive == 'yes'")

print("1. ", result.shape)
print("2. ", len(result_query))
```


## 9. 정렬

1. `pclass` 오름차순, 같은 등급 안에서는 `fare` 내림차순으로 정렬해 `pclass`, `fare`, `sex`, `age` 열의 상위 3행을 출력하세요.
2. 요금 내림차순으로 정렬한 결과를 변수에 담고, `iloc[0]`과 `loc[0]`의 `fare` 값을 각각 출력하세요. 두 값이 다른 이유를 주석으로 쓰세요.    


```python
mask = titanic.sort_values(by=["pclass", "fare"], ascending=[True, False])
result = mask[["pclass", "fare", "sex", "age"]]

fare_mask = titanic.sort_values(by="fare", ascending=False)

print("1.\n", result.head(3))
print("2. iloc[0]_fare :", fare_mask["fare"].iloc[0])
print("2. loc[0]_fare :", fare_mask["fare"].loc[0])

# iloc와 loc의 출력 값이 다른 이유는 sort_values는 정렬은 진행하지만 기존 인덱스 값은 유지하기 때문에 
# loc에 대해서는 인덱스로 행열을 선택하기 때문에 titanic.csv의 0행의 fare 값인 7.25가 출력되고,
# iloc는 현재 데이터의 위치번호로 행열을 선택하기 때문에 정렬이후 0행의  fare 값인 512.3292가 출력됩니다.
```


## 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
mask = (titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"].isin([1, 2]))
result = titanic[mask][["pclass", "age", "fare", "survived"]]
survive_result = round(result["survived"].mean(), 2)
fare_result = result.sort_values(by="fare", ascending=False)

query = titanic.query("sex == 'male' and 20 <= age <= 40 and pclass in [1, 2]")

print("row :",len(result))
print("survive :", survive_result)
print("fare top 3 :\n", fare_result.head(3))

print("\nquery row :", len(query))
print("row == query_row? :", len(result) == len(query))

pclass3_mask = (titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"].isin([3]))
pclass3_result = titanic[pclass3_mask][["pclass", "age", "fare", "survived"]]
pclass3_survive_result = round(pclass3_result["survived"].mean(), 2)

print("\npclass3_row :",len(pclass3_result))
print("pclass_3_survive :", pclass3_survive_result)
```

* 20 ~ 40세의 승객 중 1, 2등석에 앉은 사람의 생존률은 24%이고, 3등석에 앉은 사람의 생존률은 15% 입니다. 
* 1, 2등석에 앉은 사람이 3등석에 앉은 사람보다 생존률이 9%더 높았으며 좌석 등급별 생존률이 차이가 있다는 결과를 확인할 수 있습니다.
